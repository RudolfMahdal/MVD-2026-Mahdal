```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from sklearn.manifold import TSNE

values = [1, 2, 3]
x_labels = ['A', 'B', 'C'] 

plt.figure(figsize=(6, 4))
plt.bar(x_labels, values, color='skyblue')
plt.title("Matplotlib: Sloupcový graf")
plt.show()

fig1 = px.bar(x=x_labels, y=values, title="Plotly: Sloupcový graf")
fig1.show()
```

```python
with open('data.pk', 'rb') as f:
    data, Y = pickle.load(f)

data = np.array(data)
Y = np.array(Y)

# Redukce dimenze dat do 2D
tsne = TSNE(n_components=2, perplexity=5, random_state=42)
data_2d = tsne.fit_transform(data)

fig2 = px.scatter(
    x=data_2d[:, 0], 
    y=data_2d[:, 1], 
    color=Y.astype(str),
    title="t-SNE vizualizace x-vektorů",
    labels={'x': 'Dimenze 1', 'y': 'Dimenze 2', 'color': 'Osoba'}
)

fig2.update_traces(marker=dict(size=10, line=dict(width=1, color='DarkSlateGrey')))
fig2.show()
```
